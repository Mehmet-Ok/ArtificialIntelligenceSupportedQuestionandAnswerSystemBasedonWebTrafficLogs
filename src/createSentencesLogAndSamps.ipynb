```python
samp_df = pd.read_csv('/content/drive/MyDrive/logs/samp_without_query.csv')
samp_df.drop('Unnamed: 0',axis=1,inplace=True)
samp_df.head(2)
```

```python
samp_df.drop('new_request',axis=1,inplace=True)
```

```python
samp_df.drop('user_agent',axis=1,inplace=True)
```

```python
samp_df.drop('referer',axis=1,inplace=True)
```

```python
samp_df.head(2)
```

```python
samp_df["status"].value_counts()
```

```python
# Status kodlarına göre sınıflandırma
def classify_status_code(code):
    if 100 <= code < 200:
        return 'Informational'
    elif 200 <= code < 300:
        return 'Success'
    elif 300 <= code < 400:
        return 'Redirection'
    elif 400 <= code < 500:
        return 'Client Error'
    elif 500 <= code < 600:
        return 'Server Error'
    else:
        return 'Unknown'

samp_df['status_category'] = samp_df['status'].apply(classify_status_code)
```

```python
samp_df["status_category"].value_counts()
```

```python
def format_log(row):
    return (f"{row['client']} accessed {row['req_name']} on "
            f"2024-08-{row['day']}T{row['hour']:02d}:{row['minute']:02d}:00 "
            f"with {row['status_category'].lower()}.")
```

```python
print(format_log(samp_df.iloc[0]))
```

```python
sentences = samp_df.apply(format_log, axis=1)
```
